```python
# using numpy 
import numpy as np
def relu(x):
    return np.maximum(0,x)
X=np.array([[2,3]])
W=np.array([
    [1,4],
    [2,5]
])
b=np.array([1,2])
z=np.dot(X,W)+b
A=relu(z)
print(A)
```

```python
class Dense:
    def __init__(self,input_size,output_size):
        self.W=np.random.randn(input_size,output_size)
        self.b=np.zeros(output_size)
    def forward(self,X):
        return np.dot(X,self.W)+self.b
dense=Dense(3,2)
X=np.array([[1,2,3]])
output=dense.forward(X)
print(output)
```
